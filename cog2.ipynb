```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

# i
max_score = max(scores)
print("Highest score:", max_score, "at index", scores.index(max_score))

# ii
min_score = min(scores)
print("Lowest score:", min_score, "appears", scores.count(min_score), "times")

# iii
reversed_list = list(scores[::-1])
print("Reversed tuple as list:", reversed_list)

# iv
score_check = 76
if score_check in scores:
    print(f"{score_check} found at index", scores.index(score_check))
else:
    print(f"{score_check} is not present in the tuple.")
```

```python
import random

random_nums = [random.randint(100, 900) for _ in range(100)]
print("Generated numbers:", random_nums)


# i
odd_numbers = [num for num in random_nums if num % 2 != 0]
print("Odd numbers:", odd_numbers)

# ii
even_numbers = [num for num in random_nums if num % 2 == 0]
print("Even numbers:", even_numbers)

# iii
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [num for num in random_nums if is_prime(num)]
print("Prime numbers:", prime_numbers)
```

```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

# i
print("Union:", A | B)

# ii
print("Intersection:", A & B)

# iii
print("Symmetric Difference:", A ^ B)

# iv
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))

# v
X = int(input("Enter a score to remove from A: "))
if X in A:
    A.remove(X)
    print(f"{X} removed. New A:", A)
else:
    print(f"{X} is not present in set A.")
```
