```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
x=np.linspace(-10,10,10)
M = int(input("enter a value: "))
y= M* x**2
y2= M* np.sin(x)
plt.plot(y,y2, color='blue')
plt.title("functions")
plt.legend()
plt.grid(True)
plt.xlabel('x')
plt.ylabel('y')
plt.show()



df= {'subjects': ['maths', 'science', 'hindi', 'english', 'history'], 'Score': [90,99,100,100,98]}
ok= pd.DataFrame(df)

plt.figure(figsize=(8, 6))
sns.barplot(x='subjects', y='Score', data=ok, palette='Set2')

for index, row in ok.iterrows():

    plt.text(index, row['Score'] + 1,        # x and y position
             str(row['Score']),              # label
             ha='center', va='bottom', fontsize=10)

```

```python
import numpy as np
import matplotlib.pyplot as plt

roll_number = 102317026
np.random.seed(roll_number)

data = np.random.randn(50)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(np.cumsum(data), 'b-', label="Cumulative Sum")
axes[0].set_title("Line Plot of Cumulative Sum")
axes[0].set_xlabel("Index")
axes[0].set_ylabel("Cumulative Sum")
axes[0].grid(True)
axes[0].legend()

axes[1].scatter(range(50), data, c='r', marker='o', label="Random Noise")
axes[1].set_title("Scatter Plot with Random Noise")
axes[1].set_xlabel("Index")
axes[1].set_ylabel("Value")
axes[1].grid(True)
axes[1].legend()

plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
df = pd.read_csv(url)

plt.figure(figsize=(10, 5))
sns.lineplot(x=df['month_number'], y=df['total_profit'], marker='o', color='b', label="Total Profit")
plt.title("Total Profit of All Months")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(True)
plt.legend()
plt.show()


plt.figure(figsize=(10, 5))
product_columns = df.columns.drop(['month_number', 'total_profit'])
for column in product_columns:
    sns.lineplot(x=df['month_number'], y=df[column], label=column)
plt.title("Product Sales Over Months")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.legend()
plt.show()


df.set_index('month_number', inplace=True)
df.plot(kind="bar", figsize=(12, 6), width=0.8)
plt.title("Bar Chart of All Attributes")
plt.xlabel("Month")
plt.ylabel("Values")
plt.legend(loc="upper left")
plt.grid(True)
plt.tight_layout()
plt.show()
```
