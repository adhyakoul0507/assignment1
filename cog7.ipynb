```python
import numpy as np
import pandas as pd

# 1. Initialize random seed
np.random.seed(42)

# 2. Generate sales data: shape (12, 4), values between 1000 and 5000
sales_data = np.random.randint(1000, 5001, size=(12, 4))

# 3. Convert to DataFrame
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
columns = ['Electronics', 'Clothing', 'Home & Kitchen', 'Sports']
df = pd.DataFrame(sales_data, columns=columns, index=months)


print(df.head())
print(df.describe())


total_per_category = df.sum()
total_per_month = df.sum(axis=1)

avg_growth = df.pct_change().mean()


df['Total Sales'] = df.sum(axis=1)


df['Growth Rate'] = df['Total Sales'].pct_change() * 100


roll_number = 24
if roll_number % 2 == 0:
    df['Electronics'] = df['Electronics'] * 0.9
else:
    df['Clothing'] = df['Clothing'] * 0.85


import matplotlib.pyplot as plt
import seaborn as sns


df[columns].plot(kind='line', figsize=(10, 6), marker='o')
plt.title("Monthly Sales Trend")
plt.ylabel("Sales Units")
plt.xlabel("Month")
plt.grid(True)
plt.show()


plt.figure(figsize=(8, 6))
sns.boxplot(data=df[columns])
plt.title("Sales Distribution by Category")
plt.ylabel("Sales Units")
plt.show()

```
