```python
import re

text = """
Our team is working on a breakthrough quantum computing project with Dr. Smith.
The prototype is 75.8% complete but still has some issues to resolve.
For details, visit http://quantum-project.demo or email smith.j@quantum.org.
Contact support at +44 7987654321 or 555-123-4567 for immediate assistance.
We've allocated 2.5M for R&D and 1.8M for infrastructure this quarter.
Jane's team is handling the q-bit stabilization - it's quite challenging!
The project's success could revolutionize computing as we know it.
"""

def clean_text(text):
    text = re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w+\b', '', text)
    text = re.sub(r'https?://\S+', '', text)
    text = re.sub(r'\+?\d{1,3}[ -]?\d{10}', '', text)
    text = re.sub(r'\b\d{3}-\d{3}-\d{4}\b', '', text)
    return text

def custom_tokenizer(text):
    pattern = r"""
        \b\w+(?:-\w+)+\b
        | \b\d+\.\d+\b
        | \b\w+'\w+\b
        | \b\w+\b
        | [<>\w]+
    """
    return re.findall(pattern, text, re.VERBOSE)

cleaned_text = clean_text(text)
tokens = custom_tokenizer(cleaned_text)

print("Cleaned Text:\n", cleaned_text)
print("\nTokens:\n", tokens)
```

```python
import nltk
import string
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Download necessary NLTK resources
nltk.download('punkt', force=True)
nltk.download('stopwords', force=True)
# Download the 'punkt_tab' resource for sentence tokenization
nltk.download('punkt_tab')


# Original paragraph
paragraph = """i ok no yes we were sunny day outside today fun but everything is fine whatever happens happens sun fun lovely day, night okay. the moon is nice."""

# Convert to lowercase
lower_text = paragraph.lower()

# Remove punctuation
clean_text = lower_text.translate(str.maketrans('', '', string.punctuation))

# Tokenize into words and sentences
words = word_tokenize(clean_text)
sentences = sent_tokenize(clean_text)

# Display tokens
print("Word Tokenization:\n", words)
print("\nSentence Tokenization:\n", sentences)

# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

# Display filtered words
print("\nFiltered Words (No Stopwords):\n", filtered_words)

# Word frequency distribution
freq_dist = FreqDist(filtered_words)
print("\nWord Frequency Distribution:")
for word, freq in freq_dist.items():
    print(f"{word}: {freq}")
```

```python
import nltk
from nltk.stem import PorterStemmer, LancasterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords, wordnet
from nltk.tokenize import word_tokenize
import string

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

paragraph = """i ok no yes we were sunny day outside today fun but everything is fine whatever happens happens sun fun lovely day, night okay. the moon is nice."""


lower_text = paragraph.lower()
clean_text = lower_text.translate(str.maketrans('', '', string.punctuation))
words = word_tokenize(clean_text)


stop_words = set(stopwords.words("english"))
filtered_words = [word for word in words if word not in stop_words]


porter = PorterStemmer()
lancaster = LancasterStemmer()
lemmatizer = WordNetLemmatizer()

print(f"{'Word':<15}{'Porter':<15}{'Lancaster':<15}{'Lemma':<15}")
print("-" * 60)
for word in filtered_words:
    porter_stem = porter.stem(word)
    lancaster_stem = lancaster.stem(word)
    lemma = lemmatizer.lemmatize(word)
    print(f"{word:<15}{porter_stem:<15}{lancaster_stem:<15}{lemma:<15}")
```
