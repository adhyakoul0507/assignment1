```python
import numpy as np
from collections import Counter


x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
x_counts = Counter(x)
x_most_common = x_counts.most_common(1)[0][0]
x_indices = np.where(x == x_most_common)[0]

print("Most frequent value in x:", x_most_common)
print("Indices of that value in x:", x_indices)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
y_counts = Counter(y)
y_most_common = y_counts.most_common(1)[0][0]
y_indices = np.where(y == y_most_common)[0]

print("\nMost frequent value in y:", y_most_common)
print("Indices of that value in y:", y_indices)
```
