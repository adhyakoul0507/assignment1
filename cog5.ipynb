```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)

y1 = x**2
y2 = np.sin(x)
y3 = np.exp(x)
y4 = np.log(np.abs(x) + 1)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(x, y1, 'r-', label=r'')
axes[0, 0].set_title("Quadratic Function")
axes[0, 0].set_xlabel("x-axis")
axes[0, 0].set_ylabel("y-axis")
axes[0, 0].grid(True)
axes[0, 0].legend()

axes[0, 1].plot(x, y2, 'g-', label=r'')
axes[0, 1].set_title("Sine Function")
axes[0, 1].set_xlabel("x-axis")
axes[0, 1].set_ylabel("y-axis")
axes[0, 1].grid(True)
axes[0, 1].legend()

axes[1, 0].plot(x, y3, 'b-', label=r'')
axes[1, 0].set_title("Exponential Function")
axes[1, 0].set_xlabel("x-axis")
axes[1, 0].set_ylabel("y-axis")
axes[1, 0].grid(True)
axes[1, 0].legend()

axes[1, 1].plot(x, y4, 'm-', label=r'')
axes[1, 1].set_title("Logarithmic Function")
axes[1, 1].set_xlabel("x-axis")
axes[1, 1].set_ylabel("y-axis")
axes[1, 1].grid(True)
axes[1, 1].legend()

plt.tight_layout()
plt.show()


```
